### Тема 1. Моделирование геометрической вероятности
> Смоделируйте вычисление геометрической вероятности (на примере из лекции про метание дротиков в игровой круг, площадь которого в два раза меньше площади стены). Показать теоретический расчет и графическую иллюстрацию для рассматриваемого примера.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Polygon, Rectangle
from matplotlib.path import Path
```

```python
def points(n):
    a = 12
    R = np.sqrt(72 / (2 * np.sqrt(2)))
    center = a / 2

    angles = np.linspace(0, 2 * np.pi, 8, endpoint=False) + np.pi / 8
    vertices = center + R * np.column_stack([np.cos(angles), np.sin(angles)])

    pts = np.random.uniform(0, a, (n, 2))
    inside = Path(vertices).contains_points(pts)
    hits = inside.sum()
    prob = inside.mean()

    fig, ax = plt.subplots(figsize=(5, 5))
    ax.add_patch(Rectangle((0, 0), a, a, facecolor='g'))
    ax.add_patch(Polygon(vertices, facecolor='b'))
    ax.scatter(pts[:, 0], pts[:, 1], s=5, c='black')
    ax.set_aspect('equal')
    plt.show()

    print(f"В синий 8-угольник из {n} точек попало {hits}. Доля = {prob:.4f}")
```

```python
points(100)
```
