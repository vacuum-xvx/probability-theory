# Тема 2. Частотная и классическая вероятность

## Подбрасывание монеты

Для правильной монеты классическая вероятность выпадения орла равна

$$P(\text{орёл})=\frac{1}{2}=0{,}5.$$

Подбросим монету много раз и посмотрим, как меняется частота выпадения орла.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)
n_tosses = 20_000

# 1 — орёл, 0 — решка
tosses = rng.integers(0, 2, size=n_tosses)
head_frequency = np.cumsum(tosses) / np.arange(1, n_tosses + 1)
p_theory = 0.5

print(f"Классическая вероятность: {p_theory:.4f}")
print(f"Частота после {n_tosses:,} бросков: {head_frequency[-1]:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(np.arange(1, n_tosses + 1), head_frequency, color="royalblue", label="Частота орла")
ax.axhline(p_theory, color="darkorange", linestyle="--", linewidth=2, label="Вероятность = 0.5")
ax.set(title="Сходимость частоты выпадения орла", xlabel="Количество бросков", ylabel="Частота")
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```

## Вывод

При небольшом числе бросков частота заметно колеблется. С ростом числа испытаний она приближается к классической вероятности $0{,}5$.
