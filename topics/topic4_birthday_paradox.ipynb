# Тема 4. Парадокс дней рождения

Кажется маловероятным, что в небольшой группе у двух людей совпадут дни рождения. Однако уже для 23 человек эта вероятность превышает 50%.

Считаем, что в году 365 дней, все дни рождения равновероятны, а 29 февраля не учитывается.

Сначала найдём вероятность противоположного события — все дни рождения различны:

$$P(\text{все различны})=\frac{365}{365}\cdot\frac{364}{365}\cdot\ldots\cdot\frac{365-n+1}{365}.$$

Тогда

$$P(\text{есть совпадение})=1-P(\text{все различны}).$$

```python
import math


def birthday_match_probability(group_size):
    if group_size > 365:
        return 1.0
    no_match = math.prod((365 - i) / 365 for i in range(group_size))
    return 1 - no_match


for size in (10, 20, 23, 30, 50):
    print(f"{size:2d} человек: {birthday_match_probability(size):.4f}")
```

Проверим результат для группы из 23 человек с помощью моделирования.

```python
import numpy as np

rng = np.random.default_rng(42)
n_trials = 100_000
group_size = 23

birthdays = rng.integers(0, 365, size=(n_trials, group_size))
birthdays.sort(axis=1)
has_match = np.any(np.diff(birthdays, axis=1) == 0, axis=1)

p_theory = birthday_match_probability(group_size)
p_experiment = has_match.mean()
print(f"Теоретическая вероятность: {p_theory:.4f}")
print(f"Частота в моделировании:   {p_experiment:.4f}")
```

```python
import matplotlib.pyplot as plt

sizes = np.arange(2, 61)
probabilities = [birthday_match_probability(int(size)) for size in sizes]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(sizes, probabilities, color="teal", linewidth=2)
ax.axhline(0.5, color="darkorange", linestyle="--", label="Вероятность 50%")
ax.axvline(23, color="slategray", linestyle=":", label="23 человека")
ax.set(title="Вероятность совпадения дней рождения", xlabel="Размер группы", ylabel="Вероятность")
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```

## Вывод

В группе из 23 человек рассматривается много разных пар, поэтому вероятность хотя бы одного совпадения быстро растёт. Именно несоответствие между интуитивной оценкой и расчётом называют парадоксом дней рождения.
