# Тема 5. Дискретные распределения

Рассмотрим распределения Бернулли, биномиальное и Пуассона на простых примерах.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import bernoulli, binom, poisson

rng = np.random.default_rng(42)
```

## 1. Распределение Бернулли

Один бросок правильной монеты — опыт с двумя исходами. Пусть $X=1$, если выпал орёл, и $X=0$, если выпала решка.

$$P(X=1)=0{,}5,\quad P(X=0)=0{,}5.$$

```python
p = 0.5
sample = bernoulli.rvs(p, size=10_000, random_state=rng)
print(f"Теоретическое среднее: {p:.3f}")
print(f"Среднее по выборке:    {sample.mean():.3f}")
```

## 2. Биномиальное распределение

Подбросим монету 10 раз. Число орлов $Y$ имеет биномиальное распределение $Bin(10,0{,}5)$:

$$P(Y=k)=C_{10}^{k}0{,}5^k0{,}5^{10-k}.$$

```python
n = 10
k = np.arange(n + 1)
theoretical = binom.pmf(k, n, p)
experiment = rng.binomial(n, p, size=20_000)
frequencies = np.bincount(experiment, minlength=n + 1) / len(experiment)

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(k - 0.18, theoretical, width=0.36, label="Теория", color="royalblue")
ax.bar(k + 0.18, frequencies, width=0.36, label="Моделирование", color="orange")
ax.set(title="Число орлов в 10 бросках", xlabel="Количество орлов", ylabel="Вероятность", xticks=k)
ax.grid(axis="y", alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```

## 3. Распределение Пуассона

В справочную службу в среднем поступает 3 звонка в минуту. Число звонков $Z$ за минуту описывается распределением Пуассона с параметром $\lambda=3$:

$$P(Z=k)=\frac{3^k e^{-3}}{k!}.$$

```python
lam = 3
calls = np.arange(0, 12)
probabilities = poisson.pmf(calls, lam)
experiment = rng.poisson(lam, size=20_000)

print(f"Теоретическое среднее: {lam:.3f}")
print(f"Среднее по выборке:    {experiment.mean():.3f}")

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(calls, probabilities, color="teal")
ax.set(title="Число звонков за минуту", xlabel="Количество звонков", ylabel="Вероятность", xticks=calls)
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
plt.show()
```

## Вывод

- Бернулли описывает один опыт с двумя исходами.
- Биномиальное распределение показывает число успехов в фиксированной серии опытов.
- Распределение Пуассона показывает число событий за заданный промежуток времени.
